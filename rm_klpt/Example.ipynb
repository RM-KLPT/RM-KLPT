```python
x = polygen(QQ)
K.<aK> = NumberField(x^2 - 5)
p = (2^110).next_prime()
B.<i,j,k> = QuaternionAlgebra(QQ, -1, -p)
O0 = B.maximal_order(order_basis=(B.one(), i, 1/2 * (i + j), 1/2 * (1 + k)))
a = standard_RM(K, O0)
ideal = a.O_BK.random_ideal()
```

```python
result, factor = ideal.klpt(2, strong_approximation_exponent=1.60)
norm_facto = result.norm().factor()
print(f"We found an quaternion ideal of norm {result.norm().factor()}.")
print(f"If the norm is a power of 2, its size is abound p^{log(2**norm_facto[0][1], p).n(16)}.")
print(f"We check that the output ideal is equivalent to the input as left ideals, via the equation result = ideal * factor: {result == ideal * factor}.")
```
